```python
import pandas as pd

import numpy as np



def login_table(id_name_verified, id_password):

    

    Modifies the DataFrame in-place to remove 'Verified' 

    and add 'Password' from the NumPy array.

    



    id_name_verified.drop(columns=['Verified'], inplace=True)

    



    id_name_verified['Password'] = id_password[:, 1]





df = pd.DataFrame({

    'Id': [1, 2],

    'Login': ['sara', 'talha'],

    'Verified': [True, False]

})



np_array = np.array([

    [1, 'hash_abc'],

    [2, 'hash_xyz']

])



login_table(df, np_array)

print(df)
```
