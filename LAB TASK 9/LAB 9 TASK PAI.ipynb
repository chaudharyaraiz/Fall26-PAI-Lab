```python
# Import libraries
import pandas as pd
import numpy as np
import re

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report

# Sample dataset (you can replace with CSV file later)
data = {
    "text": [
        "I love this product",
        "This is the worst thing ever",
        "Amazing experience",
        "I hate it",
        "Very good quality",
        "Not worth the money",
        "Absolutely fantastic",
        "Terrible service"
    ],
    "label": [1, 0, 1, 0, 1, 0, 1, 0]  # 1 = Positive, 0 = Negative
}

df = pd.DataFrame(data)

# Text Cleaning Function
def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z]', ' ', text)  # remove special chars
    return text

df["cleaned"] = df["text"].apply(clean_text)

# Train-Test Split
X_train, X_test, y_train, y_test = train_test_split(
    df["cleaned"], df["label"], test_size=0.25, random_state=42
)

# TF-IDF Vectorization
vectorizer = TfidfVectorizer(stop_words='english')
X_train_vec = vectorizer.fit_transform(X_train)
X_test_vec = vectorizer.transform(X_test)

# Model Training
model = LogisticRegression()
model.fit(X_train_vec, y_train)

# Predictions
y_pred = model.predict(X_test_vec)

# Evaluation
print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nReport:\n", classification_report(y_test, y_pred))

# Test on new sentence
new_text = ["I really love this amazing product"]
new_vec = vectorizer.transform(new_text)
prediction = model.predict(new_vec)

print("\nPrediction:", "Positive" if prediction[0] == 1 else "Negative")
```
